# Ranges

A *range* is a Polars Series of numbers in increasing or decreasing order, each separated by a regular interval. 
Ranges are useful in a surprisingly large number of situations, so it's worthwhile to learn about them.

Integer ranges are defined using `pl.int_range`, which takes either one, two, or three arguments: a start, an end, and a step. Pass `eager=True` so the result is a Polars Series.

If you pass one argument to `pl.int_range`, this becomes the `end` value, with `start=0`, `step=1` assumed.  Two arguments give the `start` and `end` with `step=1` assumed.  Three arguments give the `start`, `end` and `step` explicitly.

A range always includes its `start` value, but does not include its `end` value.  It counts up by `step`, and it stops before it gets to the `end`.

    pl.int_range(end, eager=True): A Polars Series starting with 0 of increasing consecutive integers, stopping before end.

```python
import polars as pl
path_data = '../../../assets/data/'
```

```python
# POLARS_CHANGE: an eager range is named "literal" unless alias gives it a name
pl.int_range(5, eager=True).alias("range")
```

Notice how the Polars Series starts at 0 and goes only up to 4, not to the end value of 5.


    pl.int_range(start, end, eager=True): A Polars Series of consecutive increasing integers from start, stopping before end.

```python
pl.int_range(3, 9, eager=True).alias("range")
```


    pl.int_range(start, end, step, eager=True): A range with a difference of step between each pair of consecutive values, starting from start and stopping before end.

```python
pl.int_range(3, 30, 5, eager=True).alias("range")
```

This Polars Series starts at 3, then takes a step of 5 to get to 8, then another step of 5 to get to 13, and so on.

When you specify a step, the start, end, and step can be positive or negative whole numbers. `pl.int_range` does not take a fractional step. A fractional step is `pl.linear_space`, which takes the number of samples instead of a step. `closed="left"` includes the start and excludes the end.

```python
# POLARS_CHANGE: 7 samples from 1.5 down to but not including -2
pl.linear_space(1.5, -2, 7, closed="left", eager=True).alias("range")
```

## Example: Leibniz's formula for $\pi$

The great German mathematician and philosopher [Gottfried Wilhelm Leibniz](https://en.wikipedia.org/wiki/Gottfried_Wilhelm_Leibniz) 
(1646 - 1716) discovered a wonderful formula for $\pi$ as an infinite sum of simple fractions. The formula is

$$\pi = 4 \cdot \left(1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \frac{1}{9} - \frac{1}{11} + \dots\right)$$

Though some math is needed to establish this, we can use Polars Series to convince ourselves that the formula works. Let's calculate the first 5000 terms of Leibniz's infinite sum and see if it is close to $\pi$.

$$4 \cdot \left(1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \frac{1}{9} - \frac{1}{11} + \dots - \frac{1}{9999} \right)$$

We will calculate this finite sum by adding all the positive terms first and then subtracting the sum of all the negative terms [[1]](#footnotes):

$$4 \cdot \left( \left(1 + \frac{1}{5} + \frac{1}{9} + \dots + \frac{1}{9997} \right) - \left(\frac{1}{3} + \frac{1}{7} + \frac{1}{11} + \dots + \frac{1}{9999} \right) \right)$$

The positive terms in the sum have 1, 5, 9, and so on in the denominators. The Polars Series `by_four_to_20` contains these numbers up to 17:

```python
by_four_to_20 = pl.int_range(1, 20, 4, eager=True).alias("by_four_to_20")
by_four_to_20
```

To get an accurate approximation to $\pi$, we'll use the much longer Polars Series `positive_term_denominators`.

```python
positive_term_denominators = pl.int_range(1, 10000, 4, eager=True).alias("positive_term_denominators")
positive_term_denominators
```

The positive terms we actually want to add together are just 1 over these denominators:

```python
positive_terms = 1 / positive_term_denominators
```

The negative terms have 3, 7, 11, and so on on in their denominators. This Polars Series is just 2 added to `positive_term_denominators`.

```python
negative_terms = 1 / (positive_term_denominators + 2)
```

The overall sum is

```python
4 * ( sum(positive_terms) - sum(negative_terms) )
```

This is very close to $\pi = 3.14159\dots$. Leibniz's formula is looking good!

(footnotes)=
## Footnotes
[1] Surprisingly, when we add  *infinitely* many positive and negative fractions, the order can matter! But our approximation to $\pi$ uses only a large finite number of fractions, so it's okay to add the terms in any convenient order.
